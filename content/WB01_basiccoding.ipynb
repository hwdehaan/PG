# Basic elements of coding

## Variables

Variables are composed of 2 things:
1. a name
2. a value

Likely know from a mathematical context, but we can start more general.

In language, one can think of 'today' as a variable. It always has the same name, but its value changes everyday.

```python
today = 'Sunday'
print(today)
```

```python
today = 'Monday'
print(today)
```

This is true in general for any quantity that changes with time, place, etc.

## Conditionals: if/else

"if" statements: allow us to do different things based on different conditions

```python
today = 'Monday'

if today == 'Monday':
    alarm_time = '6:00 AM'

print(f"Today is {today} and you have to get up at {alarm_time}.")
```

```python
today = 'Sunday'

if today == 'Monday':
    alarm_time = '6:00 AM'
if today == 'Sunday':
    alarm_time = '11:00 AM'

print(f"Today is {today} and you have to get up at {alarm_time}.")
```

\
Often use if/else structure

```python
age = 25

if age > 55.0:
    print('You can get a seniors discount at Shoppers Drug Mart!')
else:
    print('Pay full price, kiddo')
```

## Loops: for and while

Conditionals give us logic, but the real power of programming is that computers are great at doing things over and over and over ...

We harness this via loops, most often a $for$ loop.

```python
for i in range(0,10):
    print(i)
```

```python
sum = 0
for i in range(0,101):
    sum = sum + i
print(sum)
```

\
$while$ loops are useful for event based looping. That is, when we don't know how often we want something to happen ahead of time, but instead want to stop the loop when something else happens.

```python
import time

count = 0
start_time = 0.0
elapsed_time = 0.0

while elapsed_time < 1E-3:
    right_now = time.time()
    if count == 0:
        start_time = right_now
    else:
        elapsed_time = right_now - start_time
    print(f'wait  {count}')
    count += 1
print('')    
print('Surprise!')
```

## Arrays

We can already do a lot with what we have: variables, if/else, loops. But often we have to deal with large amounts of data such that it is not feasible to define a separate variable for each number. In these cases, we don't use a variable which has a name and a value, but instead a $container$ that has a name and many slots for values. These different values can be accessed via their $index$. We call these $arrays$ or $matrices$.

It might be helpful to think of arrays as a vertical, single column chest of drawers. We can give the chest of drawers a name, 'chester', and refer to specific drawers via their index, which can simply be counting the drawers starting from the top (or the bottom).

```python
from IPython.display import Image, display

display(Image(filename='./chester.jpg', embed=True))
```

```python
import numpy as np

chester = np.zeros(10)

for i in range(0,10):
    chester[i] = i             #note: in python (and many other languages), counting starts at ZERO, not at one

print(chester)
```

```python
this_i = 5

print(f'The value in drawer {this_i} is {chester[this_i]}')
```

```python
print(chester[10])
```

\
So this is great! Now we can store as much data as we like and easily refer to any particular value via its index.

How about 2D arrays ?!?

We can think of a wall of mailboxes in an apartment building. 



We can store this as an array with one index, but it will be easier to work with if we now have TWO indices - one indicating how far the specific mailbox is from the top and the other indicating how far from the left.

```python
from IPython.display import Image, display

display(Image(filename='./mailbox.jpg', embed=True))
```

```python
mailboxwall = np.zeros([10,10])

print(mailboxwall)
```

```python
for icol in range(0,10):
    for irow in range(0,10):
        mailboxwall[icol,irow] = icol*10 + irow

print(mailboxwall)
```

```python
print(mailboxwall[3][7])
```
